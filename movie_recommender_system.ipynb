```python
import numpy as np
import pandas as pd

#importing the datasets
movies = pd.read_csv('/content/drive/MyDrive/tmdb_5000_movies.csv')
credits = pd.read_csv('/content/drive/MyDrive/tmdb_5000_credits.csv')

movies.head(1)
```

```python
#merge the movies and credits datasets
movies = movies.merge(credits, on='title')
movies.head(1)
```

```python
#remove the columns that are not useful and keeping only needed colums
#genres
#id
#keywords
#title
#overview
#cast
#crew

movies  = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
movies.head(1)
```

```python
movies.isnull().sum()
```

```python
#dropping the rows where overview is null
movies.dropna(inplace=True)
```

```python
#ensuring there is no duplicate values
movies.duplicated().sum()
```

```python
#Preprocessing on generes field
import ast
def convert(obj):
  L=[]
  for i in ast.literal_eval(obj): #ast used as the whole dictonary of genres was in string format
    L.append(i['name'])
  return L

movies['genres'] = movies['genres'].apply(convert)
```

```python
#keywords feild processing
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
#cast field preprocessing
def convert3(obj):
  L=[]
  counter = 0
  for i in ast.literal_eval(obj):
    if  counter != 3:
      L.append(i['name'])
      counter+=1
    else:
      break
  return L

movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies.head(1)
```

```python
#director field preprocessing
def fetch_director(obj):
  L=[]
  for i in ast.literal_eval(obj):
    if  i['job'] == 'Director':
      L.append(i['name'])
      break
  return L

movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head(1)
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","")for i in x])
movies['overview'] = movies['overview'].apply(lambda x:[i.replace(" ","")for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","")for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","")for i in x])
movies.head()
```

```python
movies['tags'] =  movies['overview'] + movies['genres' ] + movies['cast'] + movies['crew']
movies.head(1)
```

```python
#new data frame
new_df = movies[['movie_id', 'title', 'tags']]
#converting the array of tags into string
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
new_df.head()
```

```python
#converting the tags value to lower case
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
new_df.head()
```

```python
#using stemming on the words
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)

new_df['tags'] = new_df['tags'].apply(stem)
```

```python
#converting the movie tags to a vector or Vectorization
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')

vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
#using cosine similarity to know the near and similar movies
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
#main function that will reccomend similar movies
def reccomend(movie):
  movie_index = new_df[new_df['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

  for i in movies_list:
    print(new_df.iloc[i[0]].title)
```
